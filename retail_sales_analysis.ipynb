## 5. Business insights
### 5.1. Highest revenue
Which product generates the highest revenue despite potentially having fewer units sold?

```python
df_sales.groupby("Product")["Units"].sum()
```

```python
df_sales.groupby("Product")["Revenue"].sum()
```

```python
revenue_by_product = df_sales.groupby("Product")["Revenue"].sum()
units_by_product = df_sales.groupby("Product")["Units"].sum()

revenue_per_unit = revenue_by_product / units_by_product
revenue_per_unit
```

Conclusion: Laptops generate the highest total revenue (7,298 €), despite having the lowest number of units sold (9). They also have the highest revenue per unit (810.89 €), making them the most valuable product in terms of revenue generation.

This suggests that the company's revenue is strongly influenced by high-value products rather than sales volume alone.

### 5.2. Largest percentage of total revenue
Which region contributes the largest percentage of total company revenue?

```python
revenue_per_region = df_sales.groupby("Region")["Revenue"].sum()
total_revenue = df_sales["Revenue"].sum()
percentage = (revenue_per_region / total_revenue) * 100
percentage
```

Conclusion: North contributes approximately 43% of the company's total revenue. This makes North the largest regional contributor to overall sales.
